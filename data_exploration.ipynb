```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# read the data
df = pd.read_csv('data/df_cleaned.csv')
df.info()
```

```python
sns.pairplot(data=df, corner=True, hue='if_renovated');
```

```python
# option 1 just using hue for sex:
sns.catplot(
    data=df,
    y='house_age',
    x='waterfront',
    col='if_renovated',
    kind='box');
```

```python
# to check for locally dominant houses

df['locally_dominant'] = df['sqft_living'] - df['sqft_living15']
sns.relplot(
    data=df,
    x='house_age',
    y='locally_dominant',
    col='if_renovated',
    hue='waterfront',
    row='grade',
    height=4,
    aspect=1.2
);
```

```python
# to check for exceptionally large houses

sns.relplot(
    data=df,
    x='house_age',
    y='sqft_living',
    col='if_renovated',
    hue='waterfront',
    row='condition',
    height=4,
    aspect=1.2
);
```
